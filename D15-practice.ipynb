```python
list1 = [1,2,4]
list2 = [1,3,4]
n1 = len(list1)
n2 = len(list2)
num = n1 + n2
opt = []
ptr1 = ptr2 = 0
for i in range(0,num):

    if list1[ptr1] <= list2[ptr2]:
        opt.append(list1[ptr1])
        ptr1 += 1
    elif list2[ptr2] <= list1[ptr1]:
        opt.append(list2[ptr2])
        ptr2 += 1

print(opt)
```

```python
list1 = [1,2,4]
list2 = [1,3,4]

ptr1 = ptr2 = 0
opt = []

while ptr1 < len(list1) and ptr2 < len(list2):
    if list1[ptr1] <= list2[ptr2]:
        opt.append(list1[ptr1])
        ptr1 += 1
    else:
        opt.append(list2[ptr2])
        ptr2 += 1

# Add remaining elements
while ptr1 < len(list1):
    opt.append(list1[ptr1])
    ptr1 += 1

while ptr2 < len(list2):
    opt.append(list2[ptr2])
    ptr2 += 1

print(opt)
```
