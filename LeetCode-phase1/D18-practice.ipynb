## Leetcode question no 35(Search Insert position)

```python
nums = [1,2,3,4,5,6,7,8,9]
target = 18
n = len(nums)-1
left, right = 0, n

while left <= right:
    mid = (left + right)//2

    if nums[mid] == target:
        print(mid)
        break
    elif nums[mid] < target:
        left = mid + 1
    else:
        right = mid - 1

print(left)
```

## Exact Solution of Leetcode.

```python
class Solution(object):
    def searchInsert(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: int
        """
        left, right = 0, len(nums)-1

        while left <= right:
            mid = (left + right)//2

            if nums[mid] == target:
                return(mid)
            elif nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        return(left)
```
