    #### Q5. find pair with given sum(Brute Force)

```python
nums = [2,4,3,5,9,12,14,35]
target = 7
for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        if nums[i] + nums[j] == target:
            print([i,j])
```

#### Q6. find three number with given sum(Brute Force)

```python
nums = [2,3,4,5,7,9,12,15,17,20]
target = 34

for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        for k in range(j+1, len(nums)):
            if nums[i] + nums[j] + nums[k] == target:
                print([i,j,k])
```

## Leetcode question number 1(Two Sum)

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """

        hashmap = {}
        for i, num in enumerate(nums):
            complement = target - num

            if complement in nums:
                return[hashmap[complement],i]
            
            hashmap[num] = i
```

## Leetcode question number 1(Two Sum) with output.

```python
class Solution(object):
    def twoSum(self, nums, target):
        """
        :type nums: List[int]
        :type target: int
        :rtype: List[int]
        """

        hashmap = {}

        for i, num in enumerate(nums):
            complement = target - num

            if complement in hashmap:
                return [hashmap[complement], i]

            hashmap[num] = i


nums = [2, 7, 5, 8]
target = 9

obj = Solution()
result = obj.twoSum(nums, target)

print(result)
```
