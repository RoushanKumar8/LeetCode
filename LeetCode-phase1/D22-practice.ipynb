## Leetcode Question no 137(Single Number ||)

```python
nums = [0,1,0,1,0,1,99]
nums.sort()
nums
n = 0
totalNums = len(nums)
while n < totalNums:
    if nums[n] != nums[n+1]:
        n +=2
    else:
        print(nums[n])
    
```

```python
nums = [0,1,0,1,0,1,99]
seen = set()

for num in nums:
    if num in seen:
        continue
    else:
        seen.add(num)
        print(num)
```

```python
nums = [2,2,3,2,3,3,4]
ones = 0
twos = 0
for num in nums:
    ones = (ones^num) & ~twos
    twos = (twos^num) & ~ones

print(ones)
```
