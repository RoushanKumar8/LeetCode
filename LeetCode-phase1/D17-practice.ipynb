### Leetcode question no 1464(Maximum product of Two Elements in Array)

```python
nums = [3,4,5,5,6]
n = len(nums)
i = j = 0
for num in nums:
    if num > i:
        j,i = i,num
    elif num > j:
        j = num
print(i,j)
print((i-1)*(j-1))
```

## First Approach

```python
class Solution(object):
    def maxProduct(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        nums.sort()
        return((nums[-1]-1) * (nums[-2]-1))
```

##### Time Complexity: O(n log n)
##### Space Complexity: O(n)

## Second and Best Approach

```python
class Solution(object):
    def maxProduct(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        n = len(nums)
        i = j = 0
        for num in nums:
            if num > i:
                j,i = i,num
            elif num > j:
                j = num

        return((i-1)*(j-1))
```

##### Time Complexity: O(n)
##### Space Complexity: O(1)
