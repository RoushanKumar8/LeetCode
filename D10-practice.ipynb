### Leetcode question no 448(Find All Numbers Disappeared in an Array)

```python
nums = [4,3,2,7,8,2,3,1]
totalNum = set(nums)
print(totalNum)
```

```python
result = []
for i in range(1, len(nums)+1):
    if i not in totalNum:
       result.append(i)
print(result)
```
