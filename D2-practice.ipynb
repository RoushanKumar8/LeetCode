#### 2. Count Characters in a String

Problem: Count the frequency of each character.

```python
s = 'banana'
count = {}

for ch in s:
    count[ch] = count.get(ch,0)+1

print(count)
```

#### 3. Compare Two Dictionaries

Problem: Check if two strings have the same character frequencies.

```python
s = 'abbaa'
t = 'baaba'

def freq(word):
    d = {}
    for ch in word:
        d[ch] = d.get(ch, 0) + 1
    
    return d

print(freq(s) == freq(t))
```

#### 4. Find First Non-Repeating Character

```python
s = "Leetcode"

count = {}
for ch in s:
    count[ch] = count.get(ch, 0) +1

for ch in s:
    if count[ch] == 1:
        print(ch)
        break

print(count)
```

## Question no 242.(Valid Anagram)

```python
def isAnagram(self,s,t):
    if s.len != t.len:
        return False
    
    count = {}
    for ch in s:
        count[ch] = count.get(ch, 0) + 1

    for ch in t:
        if ch not in count:
            return False
        
        count[ch] -= 1
        if count[ch] < 0:
            return False
        
    return True
```
