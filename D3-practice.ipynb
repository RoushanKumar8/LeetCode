### 26. LeetCode Question No 26(Remove Duplicates from sorted Array)

```python
nums = [0,1,1,2,2,2,3,4,4,4]
k=1
for i in range(1, len(nums)):
    if nums[i] != nums[i-1]:
        nums[k] = nums[i]
        k+=1

print(nums[:k])
```

```python
class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        
        if not nums:
            return 0
        
        k=1
        for i in range(1,len(nums)):
            if nums[i] != nums[i-1]:
                nums[k] = nums[i]
                k = k+1
        return k
```
