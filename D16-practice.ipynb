## Leetcode question No. 15(3 Sum)

```python
nums = [-1,0,1,2,-1,-4]
opt = set()
result = []
length = len(nums)
for i in range(length):
    for j in range(i+1, length):
        for k in range(j+1, length):
            if nums[i] + nums[j] + nums[k] == 0:
                triplet = tuple(sorted([nums[i], nums[j], nums[k]]))
                opt.add(triplet)
print(list(opt))
```
