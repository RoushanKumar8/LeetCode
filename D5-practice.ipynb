### Leetcode question no 66(Plus One)

```python
digits = [4,3,2,1]

number = 0
for i in range(0, len(digits)):
    number = number * 10 + digits[i]

number = number + 1
digits = [int(d) for d in str(number)]
print(digits)

```

```python
class Solution(object):
    def plusOne(self, digits):
        """
        :type digits: List[int]
        :rtype: List[int]
        """
        number = 0
        for i in range(0, len(digits)):
            number = number * 10 + digits[i]

        number = number + 1
        digits = [int(d) for d in str(number)]
    
        return digits
```
