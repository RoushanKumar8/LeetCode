## Leetcode Question no 283(Move zeros).

```python
nums = [0,1,0,3,12]
count = 0
for num in nums:
    if num == 0:
        nums.remove(0)
        count += 1

while count != 0:
    nums.append(0)
    count -= 1

print(nums)
```

```python
nums = [0,1,0,3,12]
l=0
for r in range(len(nums)):
    if nums[r]==0:
        continue
    else:
        nums[l],nums[r]=nums[r],nums[l]
        l+=1
print(nums)
```

```python
nums = [0,1,0,3,12]
for num in nums:
    if num == 0:
        nums.remove(num)
        nums.append(0)
print(nums)
```
