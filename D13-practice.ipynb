## Leetcode Question no 3536(Maximum product of two digit)

```python
n = 137
rem = n%10
n=n//10
print(rem)
print(n)
```

```python
n = 5392
allNum = []
while n != 0:
    rem = n%10
    n = n//10
    allNum.append(rem)
allNum.sort()
print(allNum)
length = len(allNum)
print(length)
max = allNum[length-1] * allNum[length-2]
print(max)
```

## Leetcode Solution

```python
class Solution:
    def maxProduct(self, n: int) -> int:
        largest = second = 0

        while n:
            n, digit = divmod(n, 10)

            if digit > largest:
                second = largest
                largest = digit
            elif digit > second:
                second = digit

        return largest * second
```
