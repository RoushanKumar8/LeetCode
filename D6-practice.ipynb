### LeetCode Question no 88(Merge Sorted Array)

```python
nums1 = [1,2,7,0,0,0]
nums2 = [2,5,6]

# Worst case situation
# nums1 = [4, 5, 6, 0, 0, 0]
# nums2 = [1, 2, 3]

m = 3
n = 3

i = m-1
j = n-1
k = m+n - 1
while i>=0 and j>=0:
    if nums1[i] > nums2[j]:
        nums1[k] = nums1[i]
        i -= 1
    else:
        nums1[k] = nums2[j]
        j-=1
    k-=1

while j>=0:
    nums1[k] = nums2[j]
    j -= 1
    k -= 1

print(nums1)
```
