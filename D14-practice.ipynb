## Leetcode question no 628(Maximum Product of three number)

```python
nums = [-1,-2,-3]
n = len(nums)

largest = second = third = 0
for i in range(n+1):
    if i > largest:
        third = second
        second = largest
        largest = i
    elif i > second:
        third = second
        second = i
    elif i > third:
        third = i
    
print(largest, second, third)
print(largest * second * third)
```

### Final Solution

```python
class Solution(object):
    def maximumProduct(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        nums.sort()
        product1 = nums[-1] * nums[-2] * nums[-3]
        product2 = nums[0] * nums[1] * nums[-1]

        return(max(product1, product2))
```
