## Leetcode Question no 217.

### First Try

```python
nums = [1,2,3,1]
check = 0
while check < len(nums):
    for i in range(check+1, len(nums)):
        if nums[check] == nums[i]:
            print(True)
            break
        
    check += 1
print(False)
```

### Second Try using sort function.

```python
nums = [1,2,3,4]
nums.sort()
for i in range(0, len(nums)-1):
    if nums[i] == nums[i+1]:
        print(True)
print(False)
```

### Third Try(From ChatGPT)

```python
nums = [1,2,3,1]
seen = set()

for num in nums:
    if num in seen:
        print(True)
    seen.add(num)

print(False)
```

### Fourth Try(one line answer)

```python
class Solution(object):
    def containsDuplicate(self, nums):
        """
        :type nums: List[int]
        :rtype: bool
        """
        return len(nums) != len(set(nums))
```
