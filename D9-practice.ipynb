### Leetcode question no 268(Missing Number)

```python
nums = [9,6,4,2,3,5,7,0,1,8]

for i in range(len(nums)+1):
    if i not in nums:
        print(i)
    
```

### Actual Code

```python
class Solution(object):
    def missingNumber(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        for i in range(len(nums)+1):
            if i not in nums:
                return i
```
