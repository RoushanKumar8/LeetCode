## Leetcode Question No 206.(Reverse Linked List)

```python
class Solution(object):
    def reverseList(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        new = None
        prev = head
        if head is None:
            return head
        
        while head is not None:
            head = head.next
            prev.next = new
            new = prev
            prev = head
        
        return new
```
