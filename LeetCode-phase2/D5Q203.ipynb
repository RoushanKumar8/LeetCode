### Leetcode Question no 203.(Remove Linked List Elements)

```python
class ListNode(object):
    def __init__(self, val=0):
        self.val = val
        self.next = None
```

```python
lst = ListNode(1)
lst.next = ListNode(2)
lst.next.next = ListNode(4)

find = 2
while lst is not None:
    if lst.val == find:
        print(find)
    lst = lst.next
```

### Actual Code.

```python
# Definition for singly-linked list.
# class ListNode(object):
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution(object):
    def removeElements(self, head, val):
        """
        :type head: Optional[ListNode]
        :type val: int
        :rtype: Optional[ListNode]
        """
        if head is None:
            return head

        prev = head
        temp = head.next
        while temp is not None:
            if temp.val == val:
                prev.next = temp.next
            else:
                prev = temp
            temp = temp.next
        
        if head.val == val:
            head = head.next
        
        return head
```
