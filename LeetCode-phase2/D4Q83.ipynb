```python
class ListNode(object):
    def __init__(self, val=0):
        self.val = val
        self.next = None
```

```python
lst = ListNode(1)
lst.next = ListNode(2)
lst.next.next = ListNode(4)
while lst is not None:
    print(lst.val)
    lst = lst.next
```
