## Leetcode Question no 69...(Sqrt(x))

```python
x = 6
if x < 2:
    print(x)

left = 1
right = x//2

while left <= right:
    mid = (left + right) // 2

    if mid * mid == x:
        print(mid)

    elif mid * mid < x:
        left = mid + 1

    else:
        right = mid - 1

print(right)
```

### Actual Leetcode Code

```python
def mySqrt(self, x):
        """
        :type x: int
        :rtype: int
        """
        if x < 2:
            return x
        
        left, right = 1, x//2

        while left <= right:
            mid = (left + right) // 2

            if mid * mid == x:
                return mid
            elif mid * mid < x:
                left = mid + 1
            else: 
                right = mid - 1
        
        return right
```
